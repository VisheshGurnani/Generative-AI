# Fine-Tuning GPT-2 on a Custom Text Style

## Objective

The objective of this assignment is to understand how fine-tuning can help a pretrained language model adapt to a specific writing style.

A small GPT-2 model is fine-tuned on a custom dataset of approximately 200–300 lines of consistent fictional narrative. The model is then prompted with the **same prompt used in Assignment 2** so that the base and fine-tuned outputs can be compared fairly.

## What You Need to Do

1. Use approximately 200–300 lines of text with a consistent writing style.
2. Clean the dataset by removing timestamps, usernames, unwanted formatting, and other non-language content.
3. Fine-tune the small GPT-2 model using the Hugging Face `Trainer`.
4. Use the same prompt from Assignment 2 with both the original GPT-2 model and the fine-tuned model.
5. Compare the outputs and identify at least two clear differences in style.

## Dataset Used

A **234-line original fictional narrative** is used as the custom training dataset. It follows a consistent atmospheric mystery style with short sentences, descriptive details, dialogue, suspense, and gradual scene progression.

The dataset contains only language content, with no timestamps, usernames, chat metadata, or unnecessary formatting.

## 1. Install Required Libraries

```python
import os
import random
import numpy as np
import torch

from datasets import load_dataset
from transformers import (
    GPT2Tokenizer,
    GPT2LMHeadModel,
    DataCollatorForLanguageModeling,
    Trainer,
    TrainingArguments,
    set_seed
)

SEED = 42
set_seed(SEED)
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("PyTorch:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
print("Device:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

## 3. Load GPT-2

The base model is the same **`gpt2`** model used in Assignment 2.

The original base model is kept unchanged so that its output can be compared fairly with the fine-tuned model.

```python
model_name = "gpt2"

tokenizer = GPT2Tokenizer.from_pretrained(model_name)
base_model = GPT2LMHeadModel.from_pretrained(model_name)

# GPT-2 has no dedicated padding token.
tokenizer.pad_token = tokenizer.eos_token
base_model.config.pad_token_id = tokenizer.eos_token_id

print("Base GPT-2 loaded successfully.")
```

## 4. Prepare the Training Dataset

The custom text is tokenized into fixed-length sequences.

Because GPT-2 is a causal language model, the model learns to predict the next token from the previous tokens.

```python
raw_dataset = load_dataset(
    "text",
    data_files={"train": dataset_path}
)

print(raw_dataset)

def tokenize_function(examples):
    return tokenizer(
        examples["text"],
        truncation=True,
        max_length=128,
        padding="max_length"
    )

tokenized_dataset = raw_dataset.map(
    tokenize_function,
    batched=True,
    remove_columns=["text"]
)

data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False
)

print("Training examples:", len(tokenized_dataset["train"]))
```

## 5. Fine-Tune GPT-2 Using Hugging Face Trainer

The pretrained GPT-2 model is adapted to the vocabulary, sentence structure, pacing, atmosphere, and narrative patterns found in the custom dataset.

The training settings are intentionally modest so the experiment remains practical on a student GPU or Colab GPU.

```python
fine_tuned_model = GPT2LMHeadModel.from_pretrained(model_name)

fine_tuned_model.config.pad_token_id = tokenizer.eos_token_id

training_args = TrainingArguments(
    output_dir="gpt2_finetuned_custom_style",
    num_train_epochs=3,
    per_device_train_batch_size=2,
    gradient_accumulation_steps=4,
    learning_rate=5e-5,
    weight_decay=0.01,
    logging_steps=10,
    save_strategy="no",
    report_to="none",
    fp16=torch.cuda.is_available()
)

trainer = Trainer(
    model=fine_tuned_model,
    args=training_args,
    train_dataset=tokenized_dataset["train"],
    data_collator=data_collator
)

print("Hugging Face Trainer created successfully.")
```

```python
training_result = trainer.train()

print("\nFine-tuning completed.")
print("Training loss:", training_result.training_loss)
```

## 6. Save the Fine-Tuned Model Checkpoint

```python
checkpoint_dir = "gpt2_finetuned_custom_style"

trainer.save_model(checkpoint_dir)
tokenizer.save_pretrained(checkpoint_dir)

print("Fine-tuned checkpoint saved to:", checkpoint_dir)
```

# 7. Original GPT-2 Output

## Original Assignment 2 Prompt

> **The old house at the end of the street had been empty for years, until one rainy evening**

The prompt is kept **exactly the same as Assignment 2**.

The generation settings are also kept identical for the base and fine-tuned models.

```python
original_prompt = (
    "The old house at the end of the street had been empty for years, "
    "until one rainy evening"
)

def generate_text(model, prompt, seed=42):
    set_seed(seed)

    model = model.to(device)
    model.eval()

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to(device)

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=150,
            do_sample=True,
            temperature=0.9,
            top_k=50,
            top_p=0.95,
            repetition_penalty=1.1,
            pad_token_id=tokenizer.eos_token_id
        )

    return tokenizer.decode(
        output[0],
        skip_special_tokens=True
    )


base_output = generate_text(
    base_model,
    original_prompt,
    seed=42
)

print("BASE GPT-2 OUTPUT:\n")
print(base_output)
```

# 8. Fine-Tuned GPT-2 Output

The exact same Assignment 2 prompt and generation settings are now used with the fine-tuned GPT-2 model.

```python
fine_tuned_output = generate_text(
    fine_tuned_model,
    original_prompt,
    seed=42
)

print("FINE-TUNED GPT-2 OUTPUT:\n")
print(fine_tuned_output)
```

# 9. Before-and-After Comparison

### Difference 1 — Sentence Structure

The fine-tuned model should show more of the short, descriptive sentence structure present in the custom dataset. The base model is more likely to use a broader range of sentence structures.

### Difference 2 — Atmosphere and Vocabulary

The fine-tuned output should contain more mystery-oriented and atmospheric vocabulary, such as rain, darkness, windows, silence, clocks, footsteps, shadows, and old rooms.

### Difference 3 — Pacing

The custom dataset uses gradual suspense and relatively short sentences. The fine-tuned model should therefore develop a slower, more controlled narrative rhythm.

### Difference 4 — Thematic Consistency

The base model can continue the prompt in many different narrative directions. The fine-tuned model is more likely to remain within the fictional mystery style represented by the training dataset.

The fine-tuned model is expected to differ from the base model in at least two clear ways. First, the fine-tuned output should adopt the dataset's **short, atmospheric sentence structure**, creating a more deliberate narrative rhythm. Second, it should show stronger **mystery and suspense vocabulary**, with details such as rain, darkness, old rooms, windows, clocks, footsteps, and silence appearing more naturally.

The base GPT-2 model has a broad pretrained writing style, so it may produce more general-purpose storytelling. The fine-tuned model is biased toward the patterns represented in the custom dataset. Fine-tuning does not replace the pretrained language model; it adapts its behavior toward the style seen during training.
